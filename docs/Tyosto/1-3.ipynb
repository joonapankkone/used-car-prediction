
### Vaihe 1: Liiketoiminnalliset tavoitteet

- Luoda työkalu/ennuste auton **ostohinnan määrittämiseen** ja **myyntiajan arviointiin** valitulla alueella.
- Optimoida autoliikkeen tuotto:
  - Oikea hinnoittelu autoa **ostettaessa** ja **myytäessä**
  - Myyntiajan huomiointi, minimoidaan aika jolloin pääoma on sidottu myynnissä olevaan autoon. Voitto voi olla pienempi jos myyntiaika on lyhyt.

**Projektin päätulokset**
  - Myyntihinnan ennustaminen
  - Myyntiajan ennustaminen
  - Alueellisten erojen tunnistaminen
  - Onko jokin merkki/malli joka myy paikasta riippumatta
  - Mitkä auton ominaisuudet ovat olennaisia
  - Varusteet (eli major_options?) ja niiden vaikutus myyntihintaan
  - Katse eteenpäin: Mitä tietoja myyntiliikkeiden kannattaisi jatkossa tilastoida jotta ennustaminen olisi tarkempaa?

**Tulosten mittaaminen**
- Aineiston jakaminen **opetus- ja testiaineistoon**
- Eri mittareiden käyttö mallien arviointiin:
  - Regressiomittarit hinnalle ja myyntiajalla
  - Vai olisiko myyntiajalle luokittelumittarit? No ei ehkä.
- Jos nyt oikeaa elämää miettii, niin asiakkaan kanssa tulisi varmastikin sopia seurantajakso, jossa arvioidaan mallin toimivuutta uudella datalla. 

**Asiakas**

- Myyntiliikkeen lisäksi potentiaaliset asiakkaat voisivat olla autoa myyvät tai ostavat kuluttaja-asiakkaat
